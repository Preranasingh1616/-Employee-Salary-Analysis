```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Employee_ID': range(101, 116),

    'Name': [
        'Aarav', 'Priya', 'Rahul', 'Sneha', 'Karan',
        'Ananya', 'Rohit', 'Neha', 'Vikram', 'Pooja',
        'Arjun', 'Simran', 'Kabir', 'Isha', 'Aditya'
    ],

    'Department': [
        'IT', 'HR', 'Finance', 'IT', 'Sales',
        'HR', 'Finance', 'IT', 'Sales', 'Finance',
        'IT', 'HR', 'Sales', 'Finance', 'IT'
    ],

    'Salary': [
        50000, 35000, 45000, 60000, 40000,
        38000, 48000, 55000, 42000, 52000,
        75000, 36000, 45000, 50000, 200000
    ],

    'Experience': [
        2, 1, 3, 4, 2,
        2, 4, 3, 2, 5,
        6, 1, 3, 4, 8
    ]
}

df = pd.DataFrame(data)

df
```

```python
print(df.head())

print(df.shape)

print(df.info())

print(df.describe())
```

```python
df.drop_duplicates(inplace=True)

df.isnull().sum()
```

```python
salary_array = df['Salary'].to_numpy()

print(salary_array)
```

```python
print("Average Salary:", np.mean(salary_array))

print("Maximum Salary:", np.max(salary_array))

print("Minimum Salary:", np.min(salary_array))

print("Total Salary:", np.sum(salary_array))
```

```python
hike_amount = salary_array * 0.10

new_salary = np.add(salary_array, hike_amount)

df['New_Salary'] = new_salary

df[['Name', 'Salary', 'New_Salary']]
```

```python
df['Hike_Amount'] = df['New_Salary'] - df['Salary']

df[['Name', 'Salary', 'Hike_Amount', 'New_Salary']]
```

```python
avg_salary = np.mean(df['Salary'])

high_salary = df[df['Salary'] > avg_salary]

print(high_salary)
```

```python
high_earners = df[df['Salary'] > 50000]

print(high_earners[['Name', 'Department', 'Salary']])
```

```python
Q1 = df['Salary'].quantile(0.25)

Q3 = df['Salary'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR

upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df['Salary'] < lower_limit) |
    (df['Salary'] > upper_limit)
]

print(outliers[['Name', 'Salary']])
```

```python
print("Total Outliers:", len(outliers))
```

```python
dept_avg_salary = df.groupby('Department')['Salary'].mean()

print(dept_avg_salary)
```

```python
dept_max_salary = df.groupby('Department')['Salary'].max()

print(dept_max_salary)
```

```python
dept_min_salary = df.groupby('Department')['Salary'].min()

print(dept_min_salary)
```

```python
dept_total_salary = df.groupby('Department')['Salary'].sum()

print(dept_total_salary)
```

```python
dept_analysis = df.groupby('Department')['Salary'].agg(
    ['mean', 'max', 'min', 'sum', 'count']
)

print(dept_analysis)
```

```python
dept_analysis = dept_analysis.sort_values(
    by='mean',
    ascending=False
)

print(dept_analysis)
```

```python
top_earners = df.nlargest(3, 'Salary')

print(top_earners[['Name', 'Department', 'Salary']])
```

```python
dept_summary = df.groupby('Department').agg(
    Average_Salary=('Salary', 'mean'),
    Total_Salary=('Salary', 'sum'),
    Employee_Count=('Employee_ID', 'count')
).reset_index()

print(dept_summary)
```

```python
final_data = pd.merge(
    df,
    dept_summary,
    on='Department',
    how='left'
)

print(final_data.head())
```

```python
print(final_data.shape)

print(final_data.columns)

print(final_data.head(10))
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    dept_summary['Department'],
    dept_summary['Average_Salary']
)

plt.title('Department-wise Average Salary')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.bar(df['Name'], df['Salary'])

plt.title('Employee Salary Comparison')
plt.xlabel('Employee Name')
plt.ylabel('Salary')

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    top_earners['Name'],
    top_earners['Salary']
)

plt.title('Top 3 Highest Paid Employees')
plt.xlabel('Employee Name')
plt.ylabel('Salary')

plt.show()
```

```python
dept_count = df['Department'].value_counts()

plt.figure(figsize=(8, 5))

plt.pie(
    dept_count,
    labels=dept_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Employee Distribution by Department')

plt.show()
```

```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Calculate KPI values
total_employees = len(df)
average_salary = df['Salary'].mean()
highest_salary = df['Salary'].max()
total_departments = df['Department'].nunique()
total_outliers = len(outliers)

# Create dashboard
fig = plt.figure(figsize=(16, 11))

fig.suptitle(
    'EMPLOYEE SALARY ANALYSIS DASHBOARD',
    fontsize=20,
    fontweight='bold',
    y=0.98
)

# KPI Cards
kpis = [
    ('Total Employees', total_employees),
    ('Average Salary', f'₹{average_salary:,.0f}'),
    ('Highest Salary', f'₹{highest_salary:,.0f}'),
    ('Departments', total_departments),
    ('Salary Outliers', total_outliers)
]

positions = [0.10, 0.30, 0.50, 0.70, 0.90]

for (label, value), x in zip(kpis, positions):
    fig.text(
        x, 0.88,
        f'{label}\n{value}',
        ha='center',
        va='center',
        fontsize=12,
        fontweight='bold',
        bbox=dict(
            boxstyle='round,pad=0.8',
            facecolor='lightblue',
            edgecolor='steelblue'
        )
    )

# Create 4 charts
ax1 = fig.add_subplot(2, 2, 1)
ax2 = fig.add_subplot(2, 2, 2)
ax3 = fig.add_subplot(2, 2, 3)
ax4 = fig.add_subplot(2, 2, 4)

# Chart 1: Department-wise Average Salary
ax1.bar(
    dept_summary['Department'],
    dept_summary['Average_Salary']
)
ax1.set_title('Department-wise Average Salary')
ax1.set_xlabel('Department')
ax1.set_ylabel('Average Salary')

# Chart 2: Employee Salary Comparison
ax2.bar(df['Name'], df['Salary'])
ax2.set_title('Employee Salary Comparison')
ax2.set_xlabel('Employee')
ax2.set_ylabel('Salary')
ax2.tick_params(axis='x', rotation=45)

# Chart 3: Top 3 Highest Paid Employees
ax3.bar(
    top_earners['Name'],
    top_earners['Salary']
)
ax3.set_title('Top 3 Highest Paid Employees')
ax3.set_xlabel('Employee')
ax3.set_ylabel('Salary')

# Chart 4: Department-wise Employee Distribution
dept_count = df['Department'].value_counts()

ax4.pie(
    dept_count,
    labels=dept_count.index,
    autopct='%1.1f%%',
    startangle=90
)
ax4.set_title('Employee Distribution by Department')

plt.tight_layout(rect=[0, 0, 1, 0.82])

# Save dashboard
plt.savefig(
    'Employee_Salary_Dashboard.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
